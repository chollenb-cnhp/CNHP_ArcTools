## Merging KIC spreadsheets with parcel analysis spreadsheet from Clark filtering 4/21/2028

```python
import pandas as pd
import os
from pathlib import Path

kic_folder = Path(r"C:/Users/chollenb/GIS/SNHS/2026/LandownerSpreadsheets")

dfs = []
for file in kic_folder.glob("*.xlsx"):
    temp = pd.read_excel(file)
    temp["source_file"] = file.name
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)

#df.to_excel(r"C:\Users\chollenb\GIS\SNHS\2026\parcel_summary_private.xlsx", index=False)

parcel_summary = pd.read_excel(r"C:\Users\chollenb\GIS\SNHS\2026\parcel_summary_Clark.xlsx")

parcel_summary = parcel_summary.merge(df, how="outer", on="Owner Name")

parcel_summary.to_excel(r"C:\Users\chollenb\GIS\SNHS\2026\parcel_summary_KIC_merge.xlsx", index=False)


```
